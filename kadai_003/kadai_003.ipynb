```python
import openpyxl
import pandas as pd
from openpyxl.styles import PatternFill

# 【Step3】データの読み込み・連結・集計

# 各ExcelファイルのSheet1からデータフレームを作成
df_2022 = pd.read_excel('2022_年間売上表.xlsx', sheet_name='Sheet1')
df_2023 = pd.read_excel('2023_年間売上表.xlsx', sheet_name='Sheet1')

# 2つのデータフレームを縦に連結
df = pd.concat([df_2022, df_2023], ignore_index=True)

# 商品と売上年ごとに金額を合計
grouped = df.groupby(['商品', '売上年']).agg({'金額（千円）': 'sum'})

# グループ化した商品と売上年を通常の列に戻す
grouped = grouped.reset_index()

# 【Step4】Excelファイルへの書き込み・ヘッダーの書式変更

# Excelファイルを作成
writer = pd.ExcelWriter('売上集計表.xlsx')

# 集計結果をExcelファイルに書き込む
grouped.to_excel(writer, sheet_name='Sheet1', index=False)

# Excelファイルを閉じる
writer.close()

# 書き出したExcelファイルを読み込む
book = openpyxl.load_workbook('売上集計表.xlsx')

# ワークシートの選択
sheet = book['Sheet1']

# 薄いグレーの塗りつぶしを作成
fill = PatternFill(
    fill_type='solid',
    fgColor='F2F2F2'
)

# ヘッダー部分（1行目）の背景色を変更
for cell in sheet[1]:
    cell.fill = fill

# Excelファイルを保存
book.save('売上集計表.xlsx')

# Excelファイルを閉じる
book.close()

# 集計結果を表示
print(grouped)
```
