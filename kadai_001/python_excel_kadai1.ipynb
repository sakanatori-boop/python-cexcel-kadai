```python
import openpyxl

from datetime import date
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment
from openpyxl.utils import get_column_letter


today = date.today()

wb = Workbook()
ws = wb.active
ws.title = "請求書"

ws["B2"] = "請求書"
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws["B7"] = "担当者名:鈴木一郎 様"

ws["F4"] = "No."
ws["G4"] = "0001"
ws["F5"] = "日付"
ws["G5"] = today
ws["G5"].number_format = "yyyy/mm/dd"

ws["B10"] = "商品名"
ws["C10"] = "数量"
ws["D10"] = "単価"
ws["E10"] = "金額"

ws["B11"] = "商品A"
ws["C11"] = 2
ws["D11"] = 10000
ws["E11"] = "=C11*D11"

ws["B12"] = "商品B"
ws["C12"] = 1
ws["D12"] = 15000
ws["E12"] = "=C12*D12"

ws["E13"] = "=SUM(E11:E12)"
ws["B15"] = "合計"
ws["E15"] = "=E13"
ws["B16"] = "消費税"
ws["E16"] = "=E15*10%"
ws["B17"] = "税込合計"
ws["E17"] = "=E15+E16"

for column, width in {"A": 3, "B": 19, "C": 10, "D": 12,
                      "E": 12, "F": 8, "G": 16}.items():
    ws.column_dimensions[column].width = width

for row in (2, 10, 15, 16, 17):
    for cell in ws[row]:
        cell.font = Font(bold=True)

for row in range(11, 18):
    for column in ("C", "D", "E"):
        ws[f"{column}{row}"].alignment = Alignment(horizontal="right")

for row in (11, 12, 13, 15, 16, 17):
    for column in ("D", "E"):
        ws[f"{column}{row}"].number_format = "#,##0"

filename = f"請求書_{today:%Y%m%d}.xlsx"
wb.save(filename)

print(f"{filename} を作成しました")
```
